```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Set random seed for reproducibility
np.random.seed(71025)

# Parameters
n_participants = 100
amount = 100  # $100
delays = [0, 7, 30, 90, 180, 365, 730]  # delays in days

# Generate random parameters for each participant
k_values = np.random.uniform(2, 4, n_participants)  # discounting rate
s_values = np.random.uniform(0.08, 0.1, n_participants)  # sensitivity parameter

# Create empty list to store all data
all_data = []

# Generate indifference points for each participant
for participant in range(n_participants):
    k = k_values[participant]
    s = s_values[participant]
    
    # Calculate indifference points using hyperboloid equation
    indifference_points = []
    for delay in delays:
        # V = A / ((1 + kD)^s)
        value = amount / ((1 + k * delay) ** s)
        value += np.random.normal(-10, 10)
        indifference_points.append(value)
    
    # Store data for this participant
    for i, delay in enumerate(delays):
        all_data.append({
            'participant_id': participant + 1,
            'k': k,
            's': s,
            'delay': delay,
            'indifference_point': indifference_points[i]
        })

# Convert to DataFrame
df = pd.DataFrame(all_data)

participant_df = df.pivot(index='participant_id', columns='delay', values='indifference_point').reset_index()
participant_df.columns = ['participant_id', 'delay_0', 'delay_7', 'delay_30', 'delay_90', 'delay_180', 'delay_365', 'delay_730']
participant_df['Amount'] = amount
participant_df['Commodity'] = 'Money Loss'
participant_df
```
